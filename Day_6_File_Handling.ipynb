# **Day 6 — File Handling & Data Storage Formats in Python**

```python
import csv
import json

#Write & Read Text File

def text_file_practice():
    # Write user input to a text file
    with open("students.txt", "w") as file:
        n = int(input("Enter number of students: "))
        for i in range(n):
            name = input(f"Enter name of student {i+1}: ")
            marks = input(f"Enter marks of {name}: ")
            file.write(f"{name},{marks}\n")

    # Read and display data from file
    print("\n📖 Student Data from 'students.txt':")
    with open("students.txt", "r") as file:
        for line in file:
            name, marks = line.strip().split(",")
            print(f"Name: {name}, Marks: {marks}")

text_file_practice()
```

```python
#CSV Practice


def csv_practice():
    # Write employee data to CSV
    employees = [
        ["Name", "Role", "Salary"],
        ["Alice", "Data Analyst", 55000],
        ["Bob", "Data Engineer", 60000],
        ["Charlie", "Manager", 75000],
    ]

    with open("employees.csv", "w", newline='') as file:
        writer = csv.writer(file)
        writer.writerows(employees)

    # Read CSV and calculate average salary
    total_salary = 0
    count = 0
    with open("employees.csv", "r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            total_salary += int(row["Salary"])
            count += 1

    print(f"\n💰 Average Salary of Employees: £{total_salary / count:.2f}")


csv_practice()
```

```python
# JSON Practice


def json_practice():
    # Create JSON data
    students = {
        "students": [
            {"name": "Alice", "marks": 85, "city": "London"},
            {"name": "Bob", "marks": 75, "city": "Manchester"},
            {"name": "Charlie", "marks": 92, "city": "Birmingham"}
        ]
    }

    # Write to JSON file
    with open("students.json", "w") as file:
        json.dump(students, file, indent=4)

    # Read and filter students with marks > 80
    with open("students.json", "r") as file:
        data = json.load(file)
        print("\n🏆 Students with marks > 80:")
        for student in data["students"]:
            if student["marks"] > 80:
                print(f"{student['name']} from {student['city']} scored {student['marks']}")

json_practice()
```

```python
#Mini Project: Student Data Manager


def student_data_manager():
    # Store student info in dictionary
    students = []
    n = int(input("\nEnter number of students for data manager: "))
    for i in range(n):
        name = input(f"Enter name of student {i+1}: ")
        marks = int(input(f"Enter marks of {name}: "))
        city = input(f"Enter city of {name}: ")
        students.append({"name": name, "marks": marks, "city": city})

    # Save to text file
    with open("student_data.txt", "w") as f:
        for s in students:
            f.write(f"{s['name']},{s['marks']},{s['city']}\n")

    # Save to CSV file
    with open("student_data.csv", "w", newline='') as f:
        writer = csv.DictWriter(f, fieldnames=["name", "marks", "city"])
        writer.writeheader()
        writer.writerows(students)

    # Save to JSON file
    with open("student_data.json", "w") as f:
        json.dump({"students": students}, f, indent=4)

    # Read and display data
    print("\n📊 Student Data Saved Successfully! Displaying below:\n")
    for s in students:
        print(f"Name: {s['name']} | Marks: {s['marks']} | City: {s['city']}")

student_data_manager()
```
