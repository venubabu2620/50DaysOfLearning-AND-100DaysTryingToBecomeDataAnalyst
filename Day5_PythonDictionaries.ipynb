# **Day 5 — Python Dictionaries & Key–Value Data**

```python
#dictionary example
student = {
    "name": "Alice",
    "age": 22,
    "course": "Data Science"
}
print(student)
```

```python
#Accessing Values
print(student["name"])     # Access using key
print(student.get("age"))  # Safe access (no error if key missing)
```

```python
#Adding and Updating Values
student["grade"] = "A"           # Add new key-value pair
student["age"] = 23              # Update existing key
print(student)
```

```python
student = {
    "name": "Alice",
    "age": 23,
    "course": "Data Science",
    "grade": "A"
}

#Removing Items
student.pop("course")       # Removes key by name
student.popitem()           # Removes last inserted item
del student["age"]        # Delete specific key
print(student)
```

```python
#Looping Through a Dictionary
student = {"name": "Alice", "age": 22, "course": "Data Science"}

# Loop through keys
for key in student:
    print(key)

# Loop through values
for value in student.values():
    print(value)

# Loop through both
for key, value in student.items():
    print(key, ":", value)
```

```python
#updating
student = {"name": "Alice", "age": 22}
student.update({"course": "Data Science", "grade": "A"})
print(student)
```

```python
#Nested Dictionaries
students = {
    "s1": {"name": "Alice", "age": 22},
    "s2": {"name": "Bob", "age": 24}
}
print(students["s1"]["name"])  # Alice
```

```python
#Unique Values Using Set
marks = {"Alice": 90, "Bob": 85, "Charlie": 90}
unique_marks = set(marks.values())
print(unique_marks)  # {90, 85}
```

# **Practice Exercises**

```python
#Practice 1: Student Marks Dictionary

# Step 1: Create a dictionary of 5 students with their marks
students = {
    "Alice": 85,
    "Bob": 90,
    "Charlie": 78,
    "David": 92,
    "Emma": 88
}

# Step 2: Print all student names and marks
print("🎓 Student Marks:")
for name, marks in students.items():
    print(f"{name}: {marks}")

# Step 3: Add one more student
students["Frank"] = 80
print("\n✅ After adding Frank:")
print(students)

# Step 4: Remove one student
students.pop("Charlie")
print("\n❌ After removing Charlie:")
print(students)

# Step 5: Print average marks
average = sum(students.values()) / len(students)
print(f"\n📊 Average Marks: {average:.2f}")
```

```python
# Practice 2: Count how many times each word appears in a sentence

# Step 1: Input sentence
sentence = "I love Python because Python is powerful"

# Step 2: Convert to lowercase and split into words
words = sentence.lower().split()

# Step 3: Create an empty dictionary to store word counts
word_count = {}

# Step 4: Loop through words and count
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Step 5: Display results
print(" Word Frequency Count:")
for word, count in word_count.items():
    print(f"{word}: {count}")
```

# **Mini Project: Student Marks Dictionary**

```python
# Student Marks Dictionary

students = {
    "Alice": 85,
    "Bob": 90,
    "Charlie": 78
}

while True:
    print("\n1. View all students")
    print("2. Add student")
    print("3. Update marks")
    print("4. Remove student")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        for name, marks in students.items():
            print(f"{name}: {marks}")

    elif choice == "2":
        name = input("Enter student name: ")
        marks = int(input("Enter marks: "))
        students[name] = marks
        print(f"{name} added successfully!")

    elif choice == "3":
        name = input("Enter student name: ")
        if name in students:
            marks = int(input("Enter new marks: "))
            students[name] = marks
            print(f"{name}'s marks updated.")
        else:
            print("Student not found!")

    elif choice == "4":
        name = input("Enter student name: ")
        if name in students:
            students.pop(name)
            print(f"{name} removed.")
        else:
            print("Student not found!")

    elif choice == "5":
        print("Exiting... 👋")
        break

    else:
        print("Invalid choice, try again.")
```
