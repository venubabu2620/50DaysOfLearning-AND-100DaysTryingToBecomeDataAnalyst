# **Day 3–4 Python Practice Exercises (Functions + Lists + Tuples + Sets)**

```python
# Function that takes two numbers and returns their sum
def add_numbers(a, b):
    return a + b

# Calling the function
result = add_numbers(10, 5)
print("Sum of two numbers:", result)
```

```python
# Function with default value
def greet(name="Data Analyst"):
    print(f"Hello, {name}! Welcome to the learning journey.")

greet()
greet("Venubabu")
```

```python
#Keyword Arguments

def display_info(name, age):
    print(f"{name} is {age} years old.")

display_info(age=25, name="John")
```

```python
#local variable
def test():
    x = 10  # local variable
    print("Inside:", x)

test()
```

```python
x = 20  # global variable

def show():
    print("Inside function:", x)

show()
print("Outside function:", x)
```

```python
#Modifying Global Variable Inside Function
count = 0

def increment():
    global count
    count += 1
    print("Count inside:", count)

increment()
print("Count outside:", count)
```

```python
# Function to return the largest number in a list
def find_max(numbers):
    return max(numbers)

nums = [12, 45, 7, 89, 23]
print("Maximum number is:", find_max(nums))
```

```python
#Indexing and Slicing Lists
# Create a list
data = [10, 20, 30, 40, 50, 60]

# Indexing
print("First element:", data[0])
print("Last element:", data[-1])

# Slicing
print("Middle elements:", data[2:5])
print("Alternate elements:", data[::2])
```

```python
#Python List Methods
# Initial list
fruits = ["apple", "banana", "cherry", "mango"]
print("Original list:", fruits)

#append() – Add an item to the end
fruits.append("orange")
print("\nAfter append:", fruits)

#insert() – Add at a specific position
fruits.insert(2, "kiwi")  # insert at index 2
print("After insert:", fruits)

#remove() – Remove by value
fruits.remove("banana")
print("After remove:", fruits)

#pop() – Remove by index (default removes last)
fruits.pop(1)  # removes the second element
print("After pop:", fruits)

#sort() – Sort the list in ascending order
fruits.sort()
print("After sort (A–Z):", fruits)

#reverse() – Reverse the list order
fruits.reverse()
print("After reverse:", fruits)

#count() – Count occurrences of an element
print("Count of 'apple':", fruits.count("apple"))

#index() – Find position of a specific element
print("Index of 'mango':", fruits.index("mango"))

#copy() – Create a shallow copy of the list
new_list = fruits.copy()
print("Copied list:", new_list)

#clear() – Remove all items from the list
fruits.clear()
print("After clear:", fruits)
```

```python
#List Comprehension Examples
# Create a list of squares from 1 to 10
squares = [x**2 for x in range(1, 11)]
print("Squares:", squares)

# List of even numbers from 1 to 20
evens = [x for x in range(1, 21) if x % 2 == 0]
print("Even numbers:", evens)
```

```python
#Tuples and Unpacking
# Tuple containing personal info
person = ("Venubabu", 25, "Data Analyst")

# Unpacking tuple
name, age, role = person
print(f"Name: {name}, Age: {age}, Role: {role}")
```

```python
#Set Operations

# Create two sets
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Perform set operations
print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("Difference:", A.difference(B))
```

# **Mini Project — Student Grade Calculator**

```python
# -----------------------------
# 🎓 Student Grade Calculator
# -----------------------------

# Step 1: Define function to calculate grade based on marks
def calculate_grade(mark):
    """Return grade based on mark"""
    if mark >= 75:
        return "Distinction"
    elif mark >= 50:
        return "Pass"
    else:
        return "Fail"

# Step 2: Store student data
students = ["Alice", "Bob", "Charlie", "David", "Eva"]
marks = [85, 45, 72, 90, 60]

# Step 3: Use a list comprehension to calculate grades
grades = [calculate_grade(mark) for mark in marks]

# Step 4: Combine data using a list of tuples
student_data = list(zip(students, marks, grades))

# Step 5: Display results
print("Name\t\tMarks\tGrade")
print("-" * 30)
for name, mark, grade in student_data:
    print(f"{name}\t\t{mark}\t{grade}")

# Step 6: Use a set to find unique grades in the class
unique_grades = set(grades)
print("\nUnique Grades:", unique_grades)
```
