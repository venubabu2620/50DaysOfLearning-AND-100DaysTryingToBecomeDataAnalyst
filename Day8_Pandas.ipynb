```python
#Accessing Columns & Rows
# Accessing columns
print(df['Name'])

# Accessing multiple columns
print(df[['Name', 'City']])

# Accessing rows using .loc and .iloc
print(df.loc[1])   # by label
print(df.iloc[0])  # by position
```

```python
#Create a DataFrame of employees (Name, Department, Salary) and filter salaries above £50,000

# Create DataFrame
data = {
    'Name': ['John', 'Sara', 'Mike', 'Emma', 'Tom'],
    'Department': ['HR', 'Finance', 'IT', 'HR', 'Finance'],
    'Salary (£)': [45000, 52000, 60000, 47000, 58000]
}

df = pd.DataFrame(data)
print("Employee Data:\n", df)

# Filter salaries above £50,000
print("\nEmployees with Salary > £50,000:\n", df[df['Salary (£)'] > 50000])
```

```python
#Mini Project – Employee Data Overview

import pandas as pd

# Create DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Ella'],
    'Department': ['HR', 'Finance', 'IT', 'HR', 'Finance'],
    'Salary (£)': [42000, 55000, 61000, 48000, 53000]
}
df = pd.DataFrame(data)

# Calculate average salary
avg_salary = df['Salary (£)'].mean()
print(f"Average Salary: £{avg_salary:.2f}")

# Filter employees earning above average
above_avg = df[df['Salary (£)'] > avg_salary]
print("\nEmployees earning above average:\n", above_avg)
```
