```python
# ----------------------------------------
# 📘 1. Variables and Data Types
# ----------------------------------------

# Creating variables
name = "Venubabu"
age = 25
height = 5.9
is_student = True

# Printing variables
print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Is Student:", is_student)

# Checking data types
print("\n--- Data Types ---")
print(type(name))
print(type(age))
print(type(height))
print(type(is_student))


```

```python
# ----------------------------------------
# 📗 2. Type Casting
# ----------------------------------------

# Converting int to float
x = 10
x_float = float(x)
print("\nInteger to Float:", x_float)

# Converting float to int
y = 10.7
y_int = int(y)
print("Float to Integer:", y_int)

# Converting number to string
num = 123
num_str = str(num)
print("Number to String:", num_str, "| Type:", type(num_str))

# Converting string to int
str_num = "45"
converted_num = int(str_num)
print("String to Integer:", converted_num, "| Type:", type(converted_num))

```

```python

# ----------------------------------------
# 📕 4. Print Function and Formatting
# ----------------------------------------

print("\n--- Using print function ---")
print("Hello, welcome to Python basics!")

# Using f-strings (modern formatting)
print(f"My name is {name}, I am {age} years old, and my height is {height} feet.")

# Using .format()
print("Name: {}, Age: {}, Height: {}".format(name, age, height))


```

```python
# ----------------------------------------
# 📘 5. Basic Operators
# ----------------------------------------

a = 10
b = 3

print("\n--- Basic Operators ---")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus (Remainder):", a % b)
print("Exponent (Power):", a ** b)

```

```python

# ----------------------------------------
# 📙 6. Taking Input from User
# ----------------------------------------

print("\n--- Taking User Input ---")
user_name = input("Enter your name: ")
user_age = int(input("Enter your age: "))
fav_tool = input("Enter your favorite data tool (Python/SQL/Power BI/Excel): ")

print(f"Hello {user_name}, you are {user_age} years old and love using {fav_tool}!")
```

```python
# 🌡️ Celsius to Fahrenheit Converter

# Step 1: Take temperature input in Celsius
celsius = float(input("Enter temperature in Celsius: "))

# Step 2: Convert Celsius to Fahrenheit
fahrenheit = (celsius * 9/5) + 32

# Step 3: Display the result
print(f"{celsius}°C is equal to {fahrenheit}°F")
```

```python
# 🌡️ Celsius to Fahrenheit Converter

# Step 1: Take temperature input in Celsius
fahrenheit = float(input("Enter temperature in fahrenheit: "))

# Step 2: Convert Fahrenheit to Celsius
celsius = (fahrenheit - 32) * 5/9

# Step 3: Display the result
print(f"{fahrenheit}°F is equal to {celsius}°C")
```
