```python
#Handling Missing Data (NaN)

import pandas as pd
import numpy as np

data = {'Name': ['John', 'Sara', 'Mike', 'Emma', 'Tom'],
        'Age': [28, np.nan, 32, 25, np.nan],
        'Salary': [50000, 54000, np.nan, 42000, 58000]}

df = pd.DataFrame(data)
print("Original Data:\n", df)

# Detect missing values
print("\nMissing Values:\n", df.isnull())

# Fill missing values
df_filled = df.fillna({'Age': df['Age'].mean(), 'Salary': df['Salary'].median()})
print("\nAfter Filling Missing Values:\n", df_filled)

# Drop rows with missing values
df_dropped = df.dropna()
print("\nAfter Dropping Missing Values:\n", df_dropped)
```

```python
#Renaming & Dropping Columns
# Rename column
df.rename(columns={'Salary': 'Annual_Salary'}, inplace=True)

# Drop a column
df_dropped_col = df.drop('Age', axis=1)
print(df_dropped_col)
```

```python
#Sorting Data
# Sort by a column
sorted_df = df.sort_values(by='Annual_Salary', ascending=False)
print(sorted_df)
```

```python
#Grouping & Aggregation (GroupBy)
# Create sample data
data = {
    'Department': ['HR', 'Finance', 'Finance', 'IT', 'IT', 'HR'],
    'Employee': ['John', 'Sara', 'Tom', 'Mike', 'Emma', 'Lucy'],
    'Salary': [40000, 50000, 55000, 60000, 62000, 42000]
}

df = pd.DataFrame(data)
print("\nOriginal Data:\n", df)

# Group by department
grouped = df.groupby('Department')['Salary'].mean()
print("\nAverage Salary by Department:\n", grouped)
```

```python
#Exporting Data
# Export cleaned data to CSV
df.to_csv("cleaned_data.csv", index=False)
print("✅ Cleaned data exported to 'cleaned_data.csv'")
```

```python
#Apply Custom Functions using apply() and lambda
# Apply a 10% discount to all products priced above 500
df['Discounted_Price'] = df['Price'].apply(lambda x: x * 0.9 if x > 500 else x)
print("\nAfter Applying Discount Function:\n", df)

# Add a new column for total sales using apply
df['Total_Sales'] = df.apply(lambda row: row['Price'] * row['Quantity'], axis=1)
print("\nAfter Calculating Total Sales:\n", df)
```

```python
#Merging and Joining DataFrames
# Create two DataFrames
df_customers = pd.DataFrame({
    'CustomerID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df_orders = pd.DataFrame({
    'OrderID': [101, 102, 103],
    'CustomerID': [1, 2, 2],
    'Amount': [1200, 300, 150]
})

# Merge based on CustomerID
merged_df = pd.merge(df_customers, df_orders, on='CustomerID', how='inner')
print("\nMerged DataFrame:\n", merged_df)
```

```python
#Practice Exercises
#Handle missing data in a dataset
#Sort the dataset by Salary in descending order.
#Group by Department


import pandas as pd
import numpy as np

# Step 1: Create dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Ella'],
    'Department': ['HR', 'Finance', 'Finance', 'IT', 'HR'],
    'Age': [25, np.nan, 28, 35, np.nan],
    'Salary': [40000, 50000, np.nan, 60000, 42000]
}
df = pd.DataFrame(data)

# Step 2: Handle missing values
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)

# Step 3: Sort by Salary
df_sorted = df.sort_values(by='Salary', ascending=False)

# Step 4: Group by Department
grouped = df_sorted.groupby('Department').agg({'Salary': ['mean', 'count']})
print("Grouped Summary:\n", grouped)

# Step 5: Export data
df_sorted.to_csv("final_data.csv", index=False)
print("✅ Final cleaned data exported to 'final_data.csv'")
```

```python
#Create a DataFrame with some missing values and fill them with mean/median

import pandas as pd
import numpy as np

# Create sample data
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Mouse'],
    'Price': [800, 25, np.nan, 200, 25],
    'Quantity': [5, np.nan, 10, 3, 8]
}
df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# Fill missing values
df['Price'].fillna(df['Price'].mean(), inplace=True)
df['Quantity'].fillna(df['Quantity'].median(), inplace=True)
print("\nAfter Filling Missing Values:\n", df)
```

```python
#Remove duplicate rows from a DataFrame
# Add a duplicate row
df.loc[5] = ['Mouse', 25, 8]
print("\nDataFrame with Duplicates:\n", df)

# Remove duplicates
df_cleaned = df.drop_duplicates()
print("\nAfter Removing Duplicates:\n", df_cleaned)
```

```python
#Group sales data by product and find total sales for each
# Add new column for total sales
df_cleaned['Total_Sales'] = df_cleaned['Price'] * df_cleaned['Quantity']

# Group by product
grouped_sales = df_cleaned.groupby('Product')['Total_Sales'].sum().reset_index()
print("\nTotal Sales by Product:\n", grouped_sales)
```

```python
#Mini Project – Sales Data Cleaner
# Step 1: Create sales data
sales_data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'South'],
    'Product': ['Laptop', 'Mouse', 'Laptop', 'Monitor', 'Laptop', 'Mouse'],
    'Price': [900, 25, 850, 200, np.nan, 25],
    'Quantity': [3, 10, np.nan, 2, 5, 8],
    'CustomerID': [1, 2, 1, 3, 2, 2]
}
sales_df = pd.DataFrame(sales_data)

# Step 2: Handle missing values and duplicates
sales_df['Price'].fillna(sales_df['Price'].mean(), inplace=True)
sales_df['Quantity'].fillna(sales_df['Quantity'].median(), inplace=True)
sales_df.drop_duplicates(inplace=True)

# Step 3: Apply custom function
sales_df['Total_Revenue'] = sales_df.apply(lambda row: row['Price'] * row['Quantity'], axis=1)

# Step 4: Create customer DataFrame
customer_df = pd.DataFrame({
    'CustomerID': [1, 2, 3],
    'Customer_Name': ['Alice', 'Bob', 'Charlie'],
    'City': ['London', 'Manchester', 'Birmingham']
})

# Step 5: Merge both DataFrames
merged_data = pd.merge(sales_df, customer_df, on='CustomerID', how='left')

# Step 6: Group and summarize
summary = merged_data.groupby(['Region', 'Product'])['Total_Revenue'].sum().reset_index()
print("\nMerged and Cleaned Data:\n", merged_data)
print("\nSales Summary by Region and Product:\n", summary)

# Step 7: Export data
merged_data.to_csv("cleaned_sales_data.csv", index=False)
summary.to_csv("sales_summary.csv", index=False)
print("\n✅ Cleaned, merged, and summarized data exported successfully!")
```
